# Task 1: Exploring and Visualizing a Simple Dataset
### Objective: Learn how to load, inspect, and visualize the Iris dataset.



```python
# 1. Zaroori libraries import karna
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# 2. Dataset load karna [cite: 13]
df = sns.load_dataset('iris')

# 3. Data ko inspect karna [cite: 14, 15]
print("--- First 5 Rows (.head()) ---")
print(df.head()) # [cite: 14]

print("\n--- Dataset Shape ---")
print(df.shape) # [cite: 14]

print("\n--- Summary Statistics (.describe()) ---")
print(df.describe()) # [cite: 15]

# 4. Graphs (Visualizations) Banana [cite: 16]
print("\n--- Generating Plots ---")

# Scatter Plot [cite: 17]
plt.figure(figsize=(5, 3))
sns.scatterplot(data=df, x='sepal_length', y='sepal_width', hue='species')
plt.title('Sepal Length vs Width')
plt.show()

# Histogram [cite: 18]
plt.figure(figsize=(5, 3))
sns.histplot(data=df, x='petal_length', kde=True, hue='species')
plt.title('Petal Length Distribution')
plt.show()

# Box Plot [cite: 19]
plt.figure(figsize=(5, 3))
sns.boxplot(data=df, x='species', y='petal_width')
plt.title('Petal Width by Species')
plt.show()
```

# Task 2: Predict Future Stock Prices (Short-Term)
### Objective: Use historical stock data to predict Apple's next day closing price.


```python
# 1. Yahoo Finance ki library install aur import karna [cite: 29]
!pip install yfinance -q
import yfinance as yf
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

# 2. Apple (AAPL) ka historical data download karna [cite: 31, 32]
print("Downloading Apple Stock Data...")
stock_data = yf.download('AAPL', start='2024-01-01', end='2026-01-01', progress=False)

# 3. Features aur Target set karna (Open, High, Low, Volume to predict Close) [cite: 33]
X_stock = stock_data[['Open', 'High', 'Low', 'Volume']]
y_stock = stock_data['Close']

# 4. Data Split karna Training aur Testing mein
X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(X_stock, y_stock, test_size=0.2, random_state=42)

# 5. Linear Regression Model train karna [cite: 34]
stock_model = LinearRegression()
stock_model.fit(X_train_s, y_train_s)

# 6. Predict karna
stock_preds = stock_model.predict(X_test_s)

# 7. Actual vs Predicted ka Graph [cite: 35]
plt.figure(figsize=(9, 4))
plt.plot(y_test_s.values, label='Actual Price', color='blue')
plt.plot(stock_preds, label='Predicted Price', color='orange', linestyle='--')
plt.title('Apple Stock Price: Actual vs Predicted')
plt.ylabel('Price ($)')
plt.legend()
plt.show()
```

# Task 6: House Price Prediction
### Objective: Predict house prices using property features.


```python
# 1. Required libraries import karna
from sklearn.datasets import fetch_california_housing
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error, root_mean_squared_error

# 2. Dataset load karna [cite: 97]
housing_data = fetch_california_housing()
X_house = housing_data.data
y_house = housing_data.target  # Prices

# 3. Data Split karna
X_train_h, X_test_h, y_train_h, y_test_h = train_test_split(X_house, y_house, test_size=0.2, random_state=42)

# 4. Gradient Boosting Regression model train karna [cite: 101]
house_model = GradientBoostingRegressor(n_estimators=100, random_state=42)
house_model.fit(X_train_h, y_train_h)

# 5. Predict karna aur Evaluate karna (MAE, RMSE) [cite: 103]
house_preds = house_model.predict(X_test_h)
mae = mean_absolute_error(y_test_h, house_preds)
rmse = root_mean_squared_error(y_test_h, house_preds)

print(f"Mean Absolute Error (MAE): {mae:.4f}")
print(f"Root Mean Squared Error (RMSE): {rmse:.4f}\n")

# 6. Graph: Actual vs Predicted Prices [cite: 102]
plt.figure(figsize=(5, 5))
plt.scatter(y_test_h, house_preds, alpha=0.3, color='green')
plt.plot([y_test_h.min(), y_test_h.max()], [y_test_h.min(), y_test_h.max()], 'k--', lw=2)
plt.xlabel('Actual Prices')
plt.ylabel('Predicted Prices')
plt.title('House Prices: Actual vs Predicted')
plt.show()
```
